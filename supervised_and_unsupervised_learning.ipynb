## Unsupervided Learning

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()
x = iris.data
x
```

## K-Means Algorithm

```python
kmn = KMeans(n_clusters=3)
kmn.fit(x)
labels = kmn.predict(x)
labels
```

```python
centers = kmn.cluster_centers_
centers
```

```python
plt.scatter(x[:,0],x[:,2],c= labels)
plt.scatter(centers[:,0],centers[:,2],marker='x',s = 150)
plt.show()
```

```python
kmn.inertia_
```

```python
y = []
for i in np.arange(1,6):
    kmn = KMeans(n_clusters=i,)
    kmn.fit(x)
    y.append(kmn.inertia_)
```

```python
plt.plot(np.arange(1,6),y,'o-')
plt.xlabel('number of clusters')
plt.ylabel('inertia')
plt.show()
```

## Heirarchical Clustering

```python
from scipy.cluster.hierarchy import linkage,dendrogram,fcluster
```

```python
hierarchical = linkage(x,method = 'complete')
dendrogram(hierarchical)
plt.show()
```

```python
labels = fcluster(hierarchical,3,criterion='distance')
labels
```

```python
plt.scatter(x[:,0],x[:,2],c = labels)
plt.show()
```

## Mean Shift Algorithm

```python
from sklearn.cluster import MeanShift
```

```python
ms = MeanShift()# try this model with bandwidth = 5, now it is on its default
ms.fit(x)
labels = ms.labels_
center = ms.cluster_centers_
```

```python
plt.scatter(x[:,0],x[:,2],c= labels)
plt.scatter(center[:,0],center[:,2],marker='x',s = 150,linewidths=5)
plt.show()
```
